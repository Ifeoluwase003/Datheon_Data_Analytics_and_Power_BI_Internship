# Exploratory Data Analysis (EDA) on Superstore Retail Sales

## 1. Introduction

This project explores the Superstore Retail Sales dataset to understand sales performance, customer purchasing patterns, product category performance, regional differences, and the relationship between discounts and profitability.

The analysis uses Python and the Pandas and Matplotlib libraries to clean, examine, analyse, and visualise the dataset.

## 2. Data Loading and Preparation

The Superstore Retail Sales dataset is loaded into Python using the Pandas library.

The original dataset contains 10,800 rows and 21 columns. During data inspection, 806 malformed rows are identified and excluded because they do not contain valid order dates and sales records.

After cleaning, 9,994 valid records remain. Eleven records have missing postal codes, which are retained because postal codes are not required for the main analysis.

The Order Date column is converted into a proper date format, and a Year-Month column is created to support monthly sales analysis.

## 3. Exploratory Data Analysis and Findings

The dataset is analysed to identify patterns and trends in sales, profitability, customer purchasing behaviour, and regional performance.

The analysis covers the following areas:

1. Monthly sales trends.
2. Sales by product category.
3. Sales by geographical region.
4. Sales by customer segment.
5. Discount and profit analysis.
6. Quantity sold by product category.
7. Profit by product category.
8. Profit by geographical region.

Charts are used to present the findings and make comparisons easier to understand.

## 4. Executive Takeaways

The exploratory data analysis reveals the following key findings:

1. **Monthly Sales Performance:** November 2018 records the highest monthly sales of 118,447.83, while February 2015 records the lowest at 4,519.89.

2. **Product Category Performance:** Technology generates the highest sales of 836,154.03 and the highest total profit of 145,454.95.

3. **Regional Performance:** The West region records the highest sales of 725,457.82 and the highest profit of 108,418.45.

4. **Customer Segment Performance:** Consumer customers contribute the highest sales of 1,161,401.34, followed by Corporate and Home Office customers.

5. **Discount and Profitability:** Higher discount levels are generally associated with lower average profits. All discount levels from 30% upward in this dataset record negative average profits.

6. **Product Quantity:** Office Supplies records the highest quantity sold at 22,906 units, despite having the lowest sales revenue among the three categories.

## 5. Conclusion

The analysis shows that the Superstore generates different levels of sales and profit across product categories, customer segments, regions, and months.

Technology contributes the highest sales and profit, while the Consumer segment accounts for the largest sales. The West region also performs strongly in both sales and profitability.

The findings further suggest that higher discount levels are generally associated with lower average profits. The business should therefore review its discount strategy and pay attention to the profitability of individual product categories and regions.

Overall, the analysis provides useful insights that can support better sales planning, pricing decisions, and business performance monitoring.
